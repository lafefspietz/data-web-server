```python
logfolder = 'O:/67203/OnWafMea/DR/Data/2025_05_01_Z_Positioners/Cooldown Logs/'
subfolders = os.listdir(logfolder)
dayfolders = []
for subfolder in subfolders:
    if subfolder[2] == '-' and subfolder[5] == '-':
        dayfolders.append(subfolder)
dayfolders
```
