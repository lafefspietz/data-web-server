```python
import os
import datetime
import json
import time
from pathlib import Path
```

```python
logfolder = 'O:/67203/OnWafMea/DR/Data/2025_05_01_Z_Positioners/Cooldown Logs/'
datetimeobject = datetime.datetime.now()
yearstring = str(datetimeobject.year)[2:4]
monthstring = str(datetimeobject.month)
if len(monthstring) == 1:
    monthstring = '0' + monthstring
daystring = str(datetimeobject.day)
if len(daystring) == 1:
    daystring = '0' + daystring
datestring = yearstring + '-' + monthstring + '-' + daystring
logfiles = os.listdir(logfolder + datestring)
```

```python
while True:
    file_path = Path(logfolder + datestring + '/' + 'CH6 T ' + datestring + '.log')
    if file_path.is_file():
        f = open(file_path)
        rawdata = f.read()
        f.close()
        datalist = rawdata.split('\n')
        jsondata = {}
        jsondata['time'] = datalist[len(datalist) - 2].split(',')[1]
        jsondata['temperature'] = float(datalist[len(datalist) - 2].split(',')[2])
        file_object = open("mxctemp.json", "w")
        file_object.write(json.dumps(jsondata))
        file_object.close()
    file_path = Path(logfolder + datestring + '/' + 'CH7 T ' + datestring + '.log')
    if file_path.is_file():       
        f = open(logfolder + datestring + '/' + 'CH7 T ' + datestring + '.log')
        rawdata = f.read()
        f.close()
        datalist = rawdata.split('\n')
        jsondata = {}
        jsondata['time'] = datalist[len(datalist) - 2].split(',')[1]
        jsondata['temperature'] = float(datalist[len(datalist) - 2].split(',')[2])
        file_object = open("mxctemp2.json", "w")
        file_object.write(json.dumps(jsondata))
        file_object.close()
    file_path = Path(logfolder + datestring + '/' + 'CH2 T ' + datestring + '.log')
    if file_path.is_file():            
        f = open(logfolder + datestring + '/' + 'CH2 T ' + datestring + '.log')
        rawdata = f.read()
        f.close()
        datalist = rawdata.split('\n')
        jsondata = {}
        jsondata['time'] = datalist[len(datalist) - 2].split(',')[1]
        jsondata['temperature'] = float(datalist[len(datalist) - 2].split(',')[2])
        file_object = open("stilltemp.json", "w")
        file_object.write(json.dumps(jsondata))
        file_object.close()

    file_path = Path(logfolder + datestring + '/' + 'CH5 T ' + datestring + '.log')
    if file_path.is_file(): 
        f = open(logfolder + datestring + '/' + 'CH5 T ' + datestring + '.log')
        rawdata = f.read()
        f.close()
        datalist = rawdata.split('\n')
        jsondata = {}
        jsondata['time'] = datalist[len(datalist) - 2].split(',')[1]
        jsondata['temperature'] = float(datalist[len(datalist) - 2].split(',')[2])
        file_object = open("4kplatetemp.json", "w")
        file_object.write(json.dumps(jsondata))
        file_object.close()
    file_path = Path(logfolder + datestring + '/' + 'CH1 T ' + datestring + '.log')
    if file_path.is_file():    
        f = open(logfolder + datestring + '/' + 'CH1 T ' + datestring + '.log')
        rawdata = f.read()
        f.close()
        datalist = rawdata.split('\n')
        jsondata = {}
        jsondata['time'] = datalist[len(datalist) - 2].split(',')[1]
        jsondata['temperature'] = float(datalist[len(datalist) - 2].split(',')[2])
        file_object = open("50kplatetemp.json", "w")
        file_object.write(json.dumps(jsondata))
        file_object.close()
    time.sleep(1)
```
